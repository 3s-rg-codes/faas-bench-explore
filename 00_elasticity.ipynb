```python
import os
import sqlite3
from concurrent.futures import ThreadPoolExecutor, as_completed
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

def transform_for_elasticity_optimized(df: pd.DataFrame) -> tuple[pd.DataFrame, int]:
    if df.empty:
        return df, 0
    df = df.reset_index(drop=True)
    min_timestamp = df["timestamp"].iloc[0]
    df["timestamp"] = df["timestamp"] - min_timestamp
    return df[df["timestamp"] <= 10], int((df["timestamp"] > 10).sum())

def _fetch_config_data(db_path: str, filter_provider: str, current_config) -> pd.DataFrame:
    config_query = """
        SELECT http_req_waiting, cpu, memory, language, provider, cold_starts, date_time, timestamp
        FROM data
        WHERE experiment = ?
            AND provider = ?
            AND date_time = ?
            AND language = ?
            AND cpu = ?
            AND memory = ?
        ORDER BY timestamp
    """
    cnx_local = sqlite3.connect(db_path, check_same_thread=False)
    try:
        return pd.read_sql_query(
            config_query,
            cnx_local,
            params=[
                "elasticity",
                filter_provider,
                current_config.date_time,
                current_config.language,
                current_config.cpu,
                current_config.memory,
            ],
        )
    finally:
        cnx_local.close()

def prepare_elasticity_data_with_indexes(db_path: str, filter_provider: str, max_workers: int | None = None) -> dict:
    configs_query = """
        SELECT DISTINCT language, cpu, memory, date_time
        FROM data
        WHERE experiment = ? AND provider = ?
    """
    cnx = sqlite3.connect(db_path, check_same_thread=False)
    try:
        df_configs = pd.read_sql_query(configs_query, cnx, params=["elasticity", filter_provider])
    finally:
        cnx.close()
    df_dict = {}
    if max_workers is None:
        max_workers = min(32, (os.cpu_count() or 4) * 2)
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {
            executor.submit(_fetch_config_data, db_path, filter_provider, current_config): current_config
            for current_config in df_configs.itertuples(index=False)
        }
        for future in as_completed(futures):
            df_config = future.result()
            if df_config.empty:
                continue
            df_transformed, _ = transform_for_elasticity_optimized(df_config)
            dict_key_name = (
                f"{filter_provider}_{df_config['language'].iloc[0]}_{df_config['cpu'].iloc[0]}_{df_config['memory'].iloc[0]}"
            )
            if dict_key_name in df_dict:
                df_dict[dict_key_name] = pd.concat([df_dict[dict_key_name], df_transformed], ignore_index=True)
            else:
                df_dict[dict_key_name] = df_transformed
    return df_dict

def _coldstart_rate_per_second(value: pd.DataFrame, seconds_limit: int = 10) -> pd.DataFrame:
    df = value.copy()
    df["sec"] = np.floor(df["timestamp"]).astype(int) + 1
    df = df[df["sec"].between(1, seconds_limit)]
    df["date_time"] = df["date_time"].str[:10]
    per_sec = (
        df.groupby(["date_time", "sec"], as_index=False)["cold_starts"]
        .mean()
        .rename(columns={"cold_starts": "cold_rate"})
    )
    per_sec["cold_rate"] = per_sec["cold_rate"] * 100
    return per_sec

def plot_elasticity_with_coldstarts(value: pd.DataFrame, seconds_limit: int = 10) -> None:
    if value.empty:
        return
    fig, ax1 = plt.subplots(figsize=(6, 4), constrained_layout=True)
    sns.lineplot(
        data=value,
        x="timestamp",
        y="http_req_waiting",
        hue="memory",
        ax=ax1,
    )
    ax1.set_ylabel("Latency (ms)")
    ax1.set_xlabel("Experiment time (sec)")
    ax2 = ax1.twinx()
    per_sec = _coldstart_rate_per_second(value, seconds_limit=seconds_limit)
    sns.boxplot(
        data=per_sec,
        x="sec",
        y="cold_rate",
        color="grey",
        fill=False,
        saturation=0.5,
        ax=ax2,
    )
    ax2.set_ylabel("Cold Start Rate (%)")
    ax2.yaxis.set_major_formatter(mticker.PercentFormatter())
    plt.show()

def plot_coldstarts_latency_correlation(value: pd.DataFrame, seconds_limit: int = 10) -> None:
    if value.empty:
        return
    df = value.copy()
    df["sec"] = np.floor(df["timestamp"]).astype(int) + 1
    df = df[df["sec"].between(1, seconds_limit)]
    df["date_time"] = df["date_time"].str[:10]
    per_sec = (
        df.groupby(["date_time", "sec"], as_index=False)
        .agg(
            cold_count=("cold_starts", "sum"),
            latency_median=("http_req_waiting", "median"),
        )
    )
    if per_sec.empty:
        return
    fig, ax = plt.subplots(figsize=(5, 4), constrained_layout=True)
    sns.regplot(
        data=per_sec,
        x="cold_count",
        y="latency_median",
        scatter_kws={"alpha": 0.6},
        line_kws={"color": "black"},
        ci=None,
        ax=ax,
    )
    ax.set_xlabel("Cold starts per second")
    ax.set_ylabel("Median latency (ms)")
    ax.set_title("Cold starts vs latency per second")
    plt.show()

providers = ["google", "awsX86", "awsARM", "oracleX86", "oracleARM", "deno", "cloudflare", "flyio", "fastly"]
db_path = "data_copy.db"

for provider in providers:
    df_dict = prepare_elasticity_data_with_indexes(db_path, provider)
    for key, value in df_dict.items():
        if value.empty:
            continue
        print(f"total length for config {key}: {len(value.index)}")
        plot_elasticity_with_coldstarts(value, seconds_limit=10)
        plot_coldstarts_latency_correlation(value, seconds_limit=10)
```

```python
# Provider-aggregated plots (line + coldstart rate %) + correlation table
import os
import sqlite3
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import json
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

export_dir = Path("figures") / "00_elasticity"
export_dir.mkdir(parents=True, exist_ok=True)

config_path = Path("config") / "provider_config.json"
with config_path.open("r", encoding="utf-8") as f:
    provider_config = json.load(f)
provider_labels = provider_config.get("provider_labels", {})

def _display_label(provider_key: str) -> str:
    if provider_key == "flyio-js":
        return "Fly (JS)"
    if provider_key == "flyio-go":
        return "Fly (Go)"
    return provider_labels.get(provider_key, provider_key)

def _safe_name(text: str) -> str:
    return text.replace(" ", "_").replace("/", "-").replace("+", "plus").lower()

def _fetch_provider_rows(db_path: str, provider: str, current_config) -> pd.DataFrame:
    query = """
        SELECT http_req_waiting, cold_starts, date_time, timestamp, provider, language
        FROM data
        WHERE experiment = ?
            AND provider = ?
            AND date_time = ?
            AND language = ?
            AND cpu = ?
            AND memory = ?
        ORDER BY timestamp
    """
    cnx_local = sqlite3.connect(db_path, check_same_thread=False)
    try:
        return pd.read_sql_query(
            query,
            cnx_local,
            params=[
                "elasticity",
                provider,
                current_config.date_time,
                current_config.language,
                current_config.cpu,
                current_config.memory,
            ],
        )
    finally:
        cnx_local.close()

def _prepare_provider_df(db_path: str, provider: str, language: str | None = None, max_workers: int | None = None) -> pd.DataFrame:
    configs_query = """
        SELECT DISTINCT language, cpu, memory, date_time
        FROM data
        WHERE experiment = ? AND provider = ?
    """
    cnx = sqlite3.connect(db_path, check_same_thread=False)
    try:
        df_configs = pd.read_sql_query(configs_query, cnx, params=["elasticity", provider])
    finally:
        cnx.close()
    if language is not None:
        df_configs = df_configs[df_configs["language"] == language]
    if df_configs.empty:
        return pd.DataFrame()
    if max_workers is None:
        max_workers = min(32, (os.cpu_count() or 4) * 2)
    frames = []
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {
            executor.submit(_fetch_provider_rows, db_path, provider, current_config): current_config
            for current_config in df_configs.itertuples(index=False)
        }
        for future in as_completed(futures):
            df = future.result()
            if df.empty:
                continue
            df = df.reset_index(drop=True)
            min_ts = df["timestamp"].iloc[0]
            df["timestamp"] = df["timestamp"] - min_ts
            df = df[df["timestamp"] <= 10]
            frames.append(df)
    if not frames:
        return pd.DataFrame()
    return pd.concat(frames, ignore_index=True)

def _prepare_provider_group_df(db_path: str, providers: list[str], language: str | None = None) -> pd.DataFrame:
    frames = []
    for provider in providers:
        df = _prepare_provider_df(db_path, provider, language=language)
        if not df.empty:
            frames.append(df)
    if not frames:
        return pd.DataFrame()
    return pd.concat(frames, ignore_index=True)

def _sec_bin(df: pd.DataFrame, seconds_limit: int = 10) -> pd.DataFrame:
    tmp = df.copy()
    tmp["sec"] = np.floor(tmp["timestamp"]).astype(int)
    return tmp[tmp["sec"].between(0, seconds_limit)]

def _coldstart_rate_per_second(df: pd.DataFrame, seconds_limit: int = 10) -> pd.DataFrame:
    tmp = _sec_bin(df, seconds_limit)
    tmp["date_time"] = tmp["date_time"].str[:10]
    per_sec = (
        tmp.groupby(["date_time", "sec"], as_index=False)["cold_starts"]
        .mean()
        .rename(columns={"cold_starts": "cold_rate"})
    )
    per_sec["cold_rate"] = per_sec["cold_rate"] * 100
    return per_sec

def plot_provider_elasticity_ax(
    df: pd.DataFrame,
    provider_key: str,
    display_label: str,
    ax1: plt.Axes,
    seconds_limit: int = 10,
    show_cold_rate: bool = True,
    ) -> None:
    if df.empty:
        ax1.set_visible(False)
        return
    tmp = _sec_bin(df, seconds_limit)
    sns.lineplot(
        data=tmp,
        x="sec",
        y="http_req_waiting",
        estimator=np.median,
        errorbar="ci",
        ax=ax1,
        color="tab:blue",
    )
    ax1.set_ylabel("Median latency (ms)")
    ax1.set_xlabel("Experiment time (sec)")
    ax1.set_title(display_label)
    ax1.tick_params(axis="both", labelsize=7)

    if not show_cold_rate:
        return

    ax2 = ax1.twinx()
    per_sec_cold = _coldstart_rate_per_second(df, seconds_limit)
    sns.boxplot(
        data=per_sec_cold,
        x="sec",
        y="cold_rate",
        color="grey",
        fill=False,
        saturation=0.5,
        ax=ax2,
    )
    ax2.set_ylabel("Cold-start rate (%)")
    ax2.yaxis.set_major_formatter(mticker.PercentFormatter())
    ax2.tick_params(axis="y", labelsize=7)

def compute_provider_correlation(df: pd.DataFrame, provider_key: str, seconds_limit: int = 10) -> float:
    if df.empty:
        return float("nan")
    tmp = _sec_bin(df, seconds_limit)
    tmp["date_time"] = tmp["date_time"].str[:10]
    per_sec = (
        tmp.groupby(["date_time", "sec"], as_index=False)
        .agg(cold_count=("cold_starts", "sum"), latency_median=("http_req_waiting", "median"))
    )
    if per_sec.empty:
        return float("nan")
    return per_sec["cold_count"].corr(per_sec["latency_median"])

provider_groups = {
    "Lambda": {"providers": ["awsX86", "awsARM"], "language": None},
    "OCF": {"providers": ["oracleX86", "oracleARM"], "language": None},
    "google": {"providers": ["google"], "language": None},
    "deno": {"providers": ["deno"], "language": None},
    "cloudflare": {"providers": ["cloudflare"], "language": None},
    "flyio-js": {"providers": ["flyio"], "language": "js"},
    "flyio-go": {"providers": ["flyio"], "language": "go"},
    "fastly": {"providers": ["fastly"], "language": None},
}
db_path = "data_copy.db"

def _load_group(provider_key: str, provider_list: list[str], language: str | None) -> tuple[str, pd.DataFrame]:
    df_provider = _prepare_provider_group_df(db_path, provider_list, language=language)
    return provider_key, df_provider

max_workers = min(8, len(provider_groups))
group_results: list[tuple[str, pd.DataFrame]] = []
with ThreadPoolExecutor(max_workers=max_workers) as executor:
    futures = {
        executor.submit(_load_group, key, cfg["providers"], cfg["language"]): key
        for key, cfg in provider_groups.items()
    }
    for future in as_completed(futures):
        group_results.append(future.result())

group_map = {key: df for key, df in group_results}
ordered_keys = list(provider_groups.keys())

# Correlation table
corr_rows = []
for provider_key in ordered_keys:
    df_provider = group_map.get(provider_key, pd.DataFrame())
    corr = compute_provider_correlation(df_provider, provider_key, seconds_limit=10)
    corr_rows.append({"Provider": _display_label(provider_key), "Pearson r": corr})
corr_df = pd.DataFrame(corr_rows).set_index("Provider")
corr_display = corr_df.round(3).astype(object).where(corr_df.notna(), "n.a.")
display(corr_display)

# Combined small-multiples figure (2 rows x 4 columns)
fig, axes = plt.subplots(2, 4, figsize=(10, 5), constrained_layout=True)
axes = axes.flatten()
for ax, provider_key in zip(axes, ordered_keys):
    df_provider = group_map.get(provider_key, pd.DataFrame())
    show_cold_rate = provider_key != "fastly"
    plot_provider_elasticity_ax(
        df_provider,
        provider_key,
        _display_label(provider_key),
        ax,
        seconds_limit=10,
        show_cold_rate=show_cold_rate,
    )

# Hide any unused axes
for ax in axes[len(ordered_keys):]:
    ax.set_visible(False)

plt.show()

# Export camera-ready version (with provider titles)
export_path = export_dir / "elasticity_latency_coldrate_all_providers.pdf"
fig.savefig(export_path, bbox_inches="tight")
```

```python
# Overlay latency + cold starts 
import os
import sqlite3
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import json
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

export_dir = Path("figures") / "00_elasticity"
export_dir.mkdir(parents=True, exist_ok=True)

config_path = Path("config") / "provider_config.json"
with config_path.open("r", encoding="utf-8") as f:
    provider_config = json.load(f)
provider_labels = provider_config.get("provider_labels", {})
provider_palette = provider_config.get("provider_palette", {})
provider_linestyles = provider_config.get("provider_linestyles", {})

# Limit configs per provider for faster iterations (set to None for full data)
max_configs_per_provider = None
random_seed = 7
fallback_to_full_data = True

# Latency axis scaling
latency_log_scale = False
latency_log_ylim = (1, 20000)

_dash_lookup = {
    "solid": (),
    "dashed": (4, 2),
    "dashdot": (4, 2, 1, 2),
    "dotted": (1, 1),
}

def _dash_from_config(value):
    if isinstance(value, list):
        return tuple(value)
    if isinstance(value, tuple):
        return value
    if isinstance(value, str):
        return _dash_lookup.get(value, ())
    return ()

def _legend_linestyle(dashes: tuple):
    return (0, dashes) if dashes else "solid"

def _display_label(provider_key: str) -> str:
    if provider_key == "flyio-js":
        return "Fly (JS)"
    if provider_key == "flyio-go":
        return "Fly (Go)"
    return provider_labels.get(provider_key, provider_key)

def _safe_name(text: str) -> str:
    return text.replace(" ", "_").replace("/", "-").replace("+", "plus").lower()

def _style_base_provider(provider_key: str) -> str:
    if provider_key == "Lambda":
        return "awsX86"
    if provider_key == "OCF":
        return "oracleX86"
    if provider_key in {"flyio-js", "flyio-go"}:
        return "flyio"
    return provider_key

def _fetch_provider_rows(db_path: str, provider: str, current_config) -> pd.DataFrame:
    query = """
        SELECT http_req_waiting, cold_starts, date_time, timestamp, provider, language
        FROM data
        WHERE experiment = ?
            AND provider = ?
            AND date_time = ?
            AND language = ?
            AND cpu = ?
            AND memory = ?
        ORDER BY timestamp
    """
    cnx_local = sqlite3.connect(db_path, check_same_thread=False)
    try:
        return pd.read_sql_query(
            query,
            cnx_local,
            params=[
                "elasticity",
                provider,
                current_config.date_time,
                current_config.language,
                current_config.cpu,
                current_config.memory,
            ],
        )
    finally:
        cnx_local.close()

def _prepare_provider_df(db_path: str, provider: str, language: str | None = None, max_workers: int | None = None) -> pd.DataFrame:
    configs_query = """
        SELECT DISTINCT language, cpu, memory, date_time
        FROM data
        WHERE experiment = ? AND provider = ?
    """
    cnx = sqlite3.connect(db_path, check_same_thread=False)
    try:
        df_configs = pd.read_sql_query(configs_query, cnx, params=["elasticity", provider])
    finally:
        cnx.close()
    if language is not None:
        df_configs = df_configs[df_configs["language"] == language]
    if df_configs.empty:
        return pd.DataFrame()

    df_configs_full = df_configs.copy()
    if max_configs_per_provider is not None:
        n = min(len(df_configs), max_configs_per_provider)
        df_configs = df_configs.sample(n=n, random_state=random_seed)

    def _load_frames(selected_configs: pd.DataFrame) -> list[pd.DataFrame]:
        if selected_configs.empty:
            return []
        if max_workers is None:
            workers = min(32, (os.cpu_count() or 4) * 2)
        else:
            workers = max_workers
        frames: list[pd.DataFrame] = []
        with ThreadPoolExecutor(max_workers=workers) as executor:
            futures = {
                executor.submit(_fetch_provider_rows, db_path, provider, current_config): current_config
                for current_config in selected_configs.itertuples(index=False)
            }
            for future in as_completed(futures):
                df = future.result()
                if df.empty:
                    continue
                df = df.reset_index(drop=True)
                min_ts = df["timestamp"].iloc[0]
                df["timestamp"] = df["timestamp"] - min_ts
                df = df[df["timestamp"] <= 10]
                frames.append(df)
        return frames

    frames = _load_frames(df_configs)
    if not frames and fallback_to_full_data and max_configs_per_provider is not None:
        frames = _load_frames(df_configs_full)
    if not frames:
        return pd.DataFrame()
    return pd.concat(frames, ignore_index=True)

def _prepare_provider_group_df(db_path: str, providers: list[str], language: str | None = None) -> pd.DataFrame:
    frames = []
    for provider in providers:
        df = _prepare_provider_df(db_path, provider, language=language)
        if not df.empty:
            frames.append(df)
    if not frames:
        return pd.DataFrame()
    return pd.concat(frames, ignore_index=True)

def _sec_bin(df: pd.DataFrame, seconds_limit: int = 10) -> pd.DataFrame:
    tmp = df.copy()
    tmp["sec"] = np.floor(tmp["timestamp"]).astype(int)
    return tmp[tmp["sec"].between(0, seconds_limit)]

def _latency_per_sec(df: pd.DataFrame, seconds_limit: int = 10) -> pd.DataFrame:
    tmp = _sec_bin(df, seconds_limit)
    if tmp.empty:
        return pd.DataFrame()
    return (
        tmp.groupby("sec", as_index=False)["http_req_waiting"]
        .median()
        .rename(columns={"http_req_waiting": "latency_median"})
    )

def _coldstarts_per_sec(df: pd.DataFrame, seconds_limit: int = 10) -> pd.DataFrame:
    tmp = _sec_bin(df, seconds_limit)
    if tmp.empty:
        return pd.DataFrame()
    tmp["date_time"] = tmp["date_time"].str[:10]
    per_day = (
        tmp.groupby(["date_time", "sec"], as_index=False)["cold_starts"]
        .sum()
        .rename(columns={"cold_starts": "cold_count"})
    )
    return (
        per_day.groupby("sec", as_index=False)["cold_count"]
        .median()
        .rename(columns={"cold_count": "cold_median"})
    )

provider_groups = {
    "Lambda": {"providers": ["awsX86"], "language": None},
    "OCF": {"providers": ["oracleX86"], "language": None},
    "google": {"providers": ["google"], "language": None},
    "deno": {"providers": ["deno"], "language": None},
    "cloudflare": {"providers": ["cloudflare"], "language": None},
    "flyio-js": {"providers": ["flyio"], "language": "js"},
    "flyio-go": {"providers": ["flyio"], "language": "go"},
    "fastly": {"providers": ["fastly"], "language": None},
}
db_path = "data_copy.db"

def _load_group(provider_key: str, provider_list: list[str], language: str | None) -> tuple[str, pd.DataFrame]:
    df_provider = _prepare_provider_group_df(db_path, provider_list, language=language)
    return provider_key, df_provider

max_workers = min(8, len(provider_groups))
group_results: list[tuple[str, pd.DataFrame]] = []
with ThreadPoolExecutor(max_workers=max_workers) as executor:
    futures = {
        executor.submit(_load_group, key, cfg["providers"], cfg["language"]): key
        for key, cfg in provider_groups.items()
    }
    for future in as_completed(futures):
        group_results.append(future.result())

group_map = {key: df for key, df in group_results}
ordered_keys = list(provider_groups.keys())

rows_latency = []
rows_cold = []
for provider_key in ordered_keys:
    df_provider = group_map.get(provider_key, pd.DataFrame())
    if df_provider.empty:
        continue
    latency_df = _latency_per_sec(df_provider, seconds_limit=10)
    cold_df = _coldstarts_per_sec(df_provider, seconds_limit=10)
    if not latency_df.empty:
        latency_df["provider"] = provider_key
        rows_latency.append(latency_df)
    if provider_key != "fastly" and not cold_df.empty:
        cold_df["provider"] = provider_key
        rows_cold.append(cold_df)

latency_plot_df = pd.concat(rows_latency, ignore_index=True) if rows_latency else pd.DataFrame()
cold_plot_df = pd.concat(rows_cold, ignore_index=True) if rows_cold else pd.DataFrame()
latency_plot_df = latency_plot_df[~latency_plot_df["provider"].isin({"flyio-js", "flyio-go"})] if not latency_plot_df.empty else latency_plot_df

providers_in_data = set()
if not latency_plot_df.empty:
    providers_in_data.update(latency_plot_df["provider"].unique().tolist())
if not cold_plot_df.empty:
    providers_in_data.update(cold_plot_df["provider"].unique().tolist())

plot_providers = [key for key in ordered_keys if key in providers_in_data]
if not plot_providers:
    print("No provider data available for plotting.")
else:
    group_colors = {}
    group_dashes = {}
    for key in plot_providers:
        base = _style_base_provider(key)
        group_colors[key] = provider_palette.get(base, "#999999")
        group_dashes[key] = _dash_from_config(provider_linestyles.get(base, "solid"))

    fig, (ax_latency, ax_cold) = plt.subplots(1, 2, figsize=(8.4, 3.2), sharex=True)

    sns.lineplot(
        data=latency_plot_df,
        x="sec",
        y="latency_median",
        hue="provider",
        style="provider",
        dashes=group_dashes,
        hue_order=plot_providers,
        style_order=plot_providers,
        palette=group_colors,
        ax=ax_latency,
        estimator=None,
        errorbar=None,
        legend=False,
    )
    ax_latency.set_title("Latency")
    ax_latency.set_xlabel("Experiment time (sec)")
    ax_latency.set_ylabel("Median latency (ms)")
    if latency_log_scale:
        ax_latency.set_yscale("log")
        if latency_log_ylim is not None:
            ax_latency.set_ylim(latency_log_ylim)

    sns.lineplot(
        data=cold_plot_df,
        x="sec",
        y="cold_median",
        hue="provider",
        style="provider",
        dashes=group_dashes,
        hue_order=plot_providers,
        style_order=plot_providers,
        palette=group_colors,
        ax=ax_cold,
        estimator=None,
        errorbar=None,
        legend=False,
    )
    ax_cold.set_title("Cold starts")
    ax_cold.set_xlabel("Experiment time (sec)")
    ax_cold.set_ylabel("Median cold starts")
    ax_cold.set_yscale("linear")

    legend_handles = [
        Line2D(
            [0],
            [0],
            color=group_colors[p],
            lw=1.8,
            linestyle=_legend_linestyle(group_dashes.get(p, ()))
        )
        for p in plot_providers
    ]
    legend_labels = [_display_label(p) for p in plot_providers]
    fig.legend(
        legend_handles,
        legend_labels,
        title="Provider",
        loc="upper center",
        bbox_to_anchor=(0.5, 1.05),
        ncol=len(plot_providers),
        frameon=False,
        handlelength=1.6,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.92])
    plt.show()

    export_path = export_dir / f"elasticity_latency_coldstarts_{_safe_name('overlay')}.pdf"
    fig.savefig(export_path, bbox_inches="tight")
    plt.close(fig)
```
