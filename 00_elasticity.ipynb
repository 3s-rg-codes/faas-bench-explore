```python
import pandas as pd
import seaborn as sns
import matplotlib
matplotlib.use("Agg") 
%matplotlib inline
import matplotlib.pyplot as plt
import argparse  
import sqlite3
from threading import Thread

cnx = sqlite3.connect('./data_copy.db')
try:
    cnx.execute('''
            CREATE INDEX IF NOT EXISTS idx_elasticity_lookup 
            ON data (experiment, provider, date_time, language, cpu, memory, timestamp)
    ''')
    cnx.commit()
    print("Index created/verified")
except Exception as e:
    print(f"Index creation note: {e}")
```

```python
import os
from concurrent.futures import ThreadPoolExecutor, as_completed

def transform_for_elasticity_optimized(df: pd.DataFrame) -> tuple[pd.DataFrame, int]:
    """
    Optimized transformation - data is already filtered and sorted by SQL query
    """
    if df.empty:
        return df, 0
        
    # Data is already sorted by timestamp from SQL ORDER BY
    df = df.reset_index(drop=True)
    
    # Calculate relative timestamp
    min_timestamp = df['timestamp'].iloc[0]
    df['timestamp'] = df['timestamp'] - min_timestamp
        
    return df[df['timestamp'] <= 10], int((df['timestamp'] > 10).sum())

def _fetch_config_data(db_path: str, filter_provider: str, current_config) -> pd.DataFrame:
    config_query = '''
        SELECT http_req_waiting, cpu, memory, language, provider, cold_starts, date_time, timestamp
        FROM data 
        WHERE experiment = ? 
            AND provider = ?
            AND date_time = ?
            AND language = ?
            AND cpu = ?
            AND memory = ?
        ORDER BY timestamp
    '''
    cnx_local = sqlite3.connect(db_path, check_same_thread=False)
    try:
        return pd.read_sql_query(
            config_query,
            cnx_local,
            params=[
                'elasticity',
                filter_provider,
                current_config.date_time,
                current_config.language,
                current_config.cpu,
                current_config.memory,
            ],
        )
    finally:
        cnx_local.close()

def prepare_elasticity_data_with_indexes(db_path: str, filter_provider: str, max_workers: int | None = None) -> dict:
    # Use parameterized queries to prevent SQL injection
    configs_query = '''
        SELECT DISTINCT language, cpu, memory, date_time 
        FROM data 
        WHERE experiment = ? AND provider = ?
    '''
    cnx = sqlite3.connect(db_path, check_same_thread=False)
    try:
        df_configs = pd.read_sql_query(configs_query, cnx, params=['elasticity', filter_provider])
    finally:
        cnx.close()
    
    print(f'Found {len(df_configs)} configurations to process')
    
    df_dict = {}
    empty_df_counter = 0
    too_high_counter = 0
    
    if max_workers is None:
        max_workers = min(32, (os.cpu_count() or 4) * 2)
    
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {
            executor.submit(_fetch_config_data, db_path, filter_provider, current_config): current_config
            for current_config in df_configs.itertuples(index=False)
        }
        
        for future in as_completed(futures):
            df_config = future.result()
            if df_config.empty:
                continue
                
            df_transformed, high = transform_for_elasticity_optimized(df_config)
            too_high_counter += high
            
            if df_transformed.isna().all().all():
                empty_df_counter += 1
            
            dict_key_name = (
                f"{filter_provider}_{df_config['language'].iloc[0]}_{df_config['cpu'].iloc[0]}_{df_config['memory'].iloc[0]}"
            )
            if dict_key_name in df_dict:
                df_dict[dict_key_name] = pd.concat([df_dict[dict_key_name], df_transformed], ignore_index=True)
            else:
                df_dict[dict_key_name] = df_transformed
    
    print(f'EMPTY COUNTER {filter_provider} = {empty_df_counter}')
    print(f'HIGH COUNTER {filter_provider} = {too_high_counter}')
            
    return df_dict

providers = ["google", "awsX86", "awsARM", "oracleX86", "oracleARM", "deno", "cloudflare", "flyio", "fastly"]
db_path = 'data_copy.db'

for provider in providers:
    # Use optimized data preparation
    df_dict = prepare_elasticity_data_with_indexes(db_path, provider)

    for key, value in df_dict.items():
        safe_name = key.replace('.', '+')
        # value.to_csv(f'./../csv/{safe_name}.csv', index=True)
        
        if value.empty:
            print(f'empty dataframe for {safe_name}, skipping...')
            continue  # Changed from return to continue
        
        print(f'total length for config {key}: {len(value.index)}')
        
        # create figure and first axis
        fig, ax1 = plt.subplots(figsize=(6,4), constrained_layout=True)
        
        # primary Y-axis
        sns.lineplot(
            data=value, 
            x='timestamp',
            y='http_req_waiting', 
            hue='memory',
            ax=ax1)
        ax1.set_ylabel("Latency (ms)")
        ax1.set_xlabel("Experiment time (sec)")
        
        # secondary Y-axis (only if show_cold=True)
        ax2 = ax1.twinx()
            
        # concat date_time to only date part
        value['date_time'] = value['date_time'].str[:10]
            
        # group by timestamp and day
        cold_avg = value.groupby(['date_time', 'timestamp'], as_index=False)['cold_starts'].sum()
            
        # cold_avg.to_csv(f'./../csv/{safe_name}.csv', index=True)
            
        # divide by number of experiments per day
        cold_avg['cold_starts'] = cold_avg['cold_starts'].where(cold_avg['cold_starts'] != 0, 0) / 96
            
        cold_avg = cold_avg.reset_index()
            
        sns.boxplot(
            data=cold_avg,
            x='timestamp',
            y='cold_starts',
            color='grey',
            fill=False,
            saturation=0.5,
            ax=ax2
        )
        ax2.set_ylabel("Cold Starts")
        
        plt.show()
```
